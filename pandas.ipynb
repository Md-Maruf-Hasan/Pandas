#Dictionary in series 

```python
import pandas as pd 
print("Dictionary  to series")
dic = { "Name": ["Maruf ", " Arif ", "Alif"], 
       "Subject": ["EEE ", "SWE", 'Math'] ,
       "Age": [21, 22, 23] }

var1 = pd.Series(dic) # Creating a series from dictionary
print(var1)
```

#Arithmetic operations

```python
import pandas as pd

Marks = pd.DataFrame({
    "Student name": ["Rafiq", "Al-Amin", "Kuddus"],
    "MATH": [80, 65, 83],
    "BANGLA": [68, 69, 69],
    "ENGLISH": [68, 57, 97],
})
Marks["TOTAL"] = Marks["MATH"] + Marks["BANGLA"] + Marks["ENGLISH"]
Marks["Percentage"] = (Marks["TOTAL"] / 300) * 100
Marks["Result"] = Marks["Percentage"].apply(
    lambda x: "Pass" if x >= 50 else "Fail"
)
print(Marks)
```

#INsert 

```python
import pandas as pd 

var = pd.DataFrame({ "value":[ 10, 21,32,321,323 ],
         "point":[1.2 ,3.5,5.5,3.3 ,5.0]})

print(var)
# Creating new column in  the current sheet
var.insert(2,"New_data",[1,22,33,44,55]) #column number, column name , data 
print( var)

# Copy from current column 
var.insert(1 , "COPIED COLUMN", var["point"]) # column number , new column name , column name that I want to copy 
print(" " , var) 

# Copy from current column but only specific number of data.

var["specific_copy"] = var["COPIED COLUMN"][0:3] # here we use sclicing
print(' ')

print(var)
```
